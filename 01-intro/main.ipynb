Q7. Sum of weights

```python
df_asia = df[df['origin'] == 'Asia']
df_asia
```

```python
df_filtered = df_asia[['vehicle_weight', 'model_year']]
df_filtered
```

```python
df_filtered.head(n=7)
```

```python
X = df_filtered.head(n=7).values
X
```

```python
X = np.array([[4240, 1996],
       [4310, 1992],
       [4230, 1997],
       [4420, 1978],
       [3950, 2007],
       [4750, 2021],
       [4450, 2019]])
X_transpose = X.T
X.T
```

```python
XTX = X.T.dot(X) 
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
w.sum(0).round(3)
```
